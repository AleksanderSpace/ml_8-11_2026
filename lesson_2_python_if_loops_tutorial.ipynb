## 1. Операторы сравнения

Операторы сравнения возвращают логическое значение: `True` (истина) или `False` (ложь).

| Оператор | Значение | Пример | Результат |
|----------|----------|--------|-----------|
| `==` | равно | `5 == 5` | `True` |
| `!=` | не равно | `5 != 5` | `False` |
| `>`  | больше | `7 > 3` | `True` |
| `<`  | меньше | `7 < 3` | `False` |
| `>=` | больше или равно | `5 >= 5` | `True` |
| `<=` | меньше или равно | `4 <= 5` | `True` |

⚠️ Не путайте `=` (присваивание) и `==` (сравнение).


````python
# Проверим операторы сравнения
a, b = 10, 5

print(a == b)   # равно?
print(a != b)   # не равно?
print(a > b)    # больше?
print(a < b)    # меньше?
print(a >= b)   # больше или равно?
print(a <= b)   # меньше или равно?
````

````python
# Сравнение работает и для строк (по алфавиту)
print("яблоко" == "яблоко")
print("ананас" < "яблоко")  # 'а' идёт раньше 'я' в алфавите -> True
print(len("код") >= 3)
````

## 2. Логические операторы: `and`, `or`, `not`

- **`and` (И)** — True, только если **оба** условия истинны.
- **`or` (ИЛИ)** — True, если **хотя бы одно** условие истинно.
- **`not` (НЕ)** — инвертирует значение.

Таблицы истинности:

| A | B | A and B | A or B |
|---|---|---------|--------|
| True | True | **True** | **True** |
| True | False | False | **True** |
| False | True | False | **True** |
| False | False | False | **False** |

| A | not A |
|---|-------|
| True | **False** |
| False | **True** |


````python
# Логические операторы в действии
x = 15

print(x > 10 and x < 20)   # 15 между 10 и 20 -> True
print(x > 20 and x < 30)   # оба условия ложны -> False
print(x > 20 or x < 30)    # второе условие истинно -> True
print(not (x > 10))        # инверсия -> False
print(14 > x > 10)
````

````python
# Пример: проверка, можно ли купить билет
age = 17
has_money = True

# Можно купить, если совершеннолетний И есть деньги
can_buy = age >= 18 and has_money
print("Можно купить билет:", can_buy)
````

````python
# Можно комбинировать несколько условий
temp = 17
is_raining = False

# Хорошая погода: тепло (>= 18) И без дождя
good_weather = temp >= 18 and not is_raining
if good_weather:
  print("Хорошая погода:")
else:
  print("bad")
````

## 3. Условия: `if`, `elif`, `else`

Условная конструкция позволяет выполнять код только при выполнении условия.

```
if условие_1:
    # выполнится, если условие_1 == True
elif условие_2:
    # выполнится, если условие_1 == False, а условие_2 == True
elif условие_3:
    # выполнится, если условие_1 == False, а условие_2 == True
else:
    # выполнится, если все условия выше == False
```

Важно: в Python блоки кода выделяются **отступами** (обычно 4 пробела).
`elif` (сокращение от else if) можно использовать несколько раз, `else` — не более одного.


````python
# Простой пример if
temperature = 25

if temperature > 20:
    print("Тепло, можно в футболке!")
````

````python
# if-else
temperature = 10

if temperature > 20:
    print("Тепло")
else:
    print("Прохладно")
````

````python
# if-elif-else: оценка за экзамен
score = 76

if score >= 90:
    grade = "5 (отлично)"
elif score >= 75:
    grade = "4 (хорошо)"
elif score >= 60:
    grade = "3 (удовлетворительно)"
else:
    grade = "2 (неудовлетворительно)"

print(f"Балл {score} -> оценка {grade}")
````

````python
# Измените score и выполните ячейку снова — убедитесь, что работают все ветки
score = 50

if score >= 75:
    grade = "5 (отлично)"
elif score >= 75:
    grade = "4 (хорошо)"
elif score >= 60:
    grade = "3 (удовлетворительно)"
else:
    grade = "2 (неудовлетворительно)"

print(f"Балл {score} -> оценка {grade}")
````

````python
# if как часть выражения (тернарный оператор)
age = 20
status = "совершеннолетний" if age >= 18 else "несовершеннолетний"
print(status)
````

## 4. Вложенные условия

Условия можно вкладывать друг в друга — внутри блока `if` может быть ещё один `if`.


````python
# Вложенные условия: доступ в систему
login = "admin"
password = "1234"

if login == "admin":
    if password == "1234":
        print("Доступ разрешён")
    else:
        print("Неверный пароль")
else:
    print("Пользователь не найден")
````

````python
# Тот же пример, но проще — через and
login = "admin"
password = "1234"

if login == "admin" and password == "1234" \
      and password == "1234" and password == "1234" \
      and password == "1234" and password == "1234":
    print("Доступ разрешён")
else:
    print("Ошибка входа")
````

## 5. Цикл `while` — повторяем, пока условие истинно

```
while условие:
    # тело цикла выполняется, пока условие == True
```

⚠️ Если условие никогда не станет ложным — получится **бесконечный цикл**!


````python
# Выводим числа от 1 до 5
i = 1
while i <= 5:
    print(i)
    i = i + 3   # увеличиваем i, иначе цикл будет бесконечным!
print("Цикл завершён")
````

````python
# while с накоплением суммы
n = 10
i = 1
total = 0

while i <= n:
    total += i     # total = total + i
    i += 1
print(f"Сумма чисел от 1 до {n} равна {total}")
````

````python
# Угадай число: классический пример цикла while
# (догадки заданы заранее, чтобы ячейка работала автоматически)
secret = 7
guess = 0
attempts = 0
guesses = [3, 9, 7]

while guess != secret:
    guess = guesses[attempts]
    attempts += 1
    if guess < secret:
        print(f"Попытка {attempts}: {guess} — слишком мало!")
    elif guess > secret:
        print(f"Попытка {attempts}: {guess} — слишком много!")

print(f"Угадали за {attempts} попыток!")
````

## 6. Цикл `for` и функция `range()`

Цикл `for` перебирает элементы **последовательности** (список, строка, range и т.д.).

- `range(стоп)` — числа от 0 до стоп-1
- `range(старт, стоп)` — от старта до стоп-1
- `range(старт, стоп, шаг)` — с заданным шагом


````python
# Перебор списка
fruits = ["яблоко", "банан", "апельсин", "груша", ("ewt",323), [[2],["f"]]]

for fruit in fruits:
    for el in fruit:
      print(el)
    print(f"Я люблю {fruit}!")
````

````python
# Перебор строки посимвольно
for letter in "Python":
    print(letter)
````

````python
# range в трёх вариантах
print("range(5):       ", list(range(5)))
print("range(2, 6):    ", list(range(2, 6)))
print("range(1, 10, 2):", list(range(1, 10, 2)))
````

````python
# Таблица умножения на 7 через for + range
for i in range(1, 11):
    print(f"7 x {i} = {7 * i}")
````

## 7. Управление циклами: `break`, `continue`, `else`

- **`break`** — немедленно выйти из цикла.
- **`continue`** — пропустить оставшуюся часть итерации и перейти к следующей.
- **`else` в цикле** — выполнится, только если цикл завершился **без** `break`.


````python
# break: выходим, как только нашли чётное
for num in [1, 3, 5, 8, 9, 10]:
     if num % 2 == 0:
        print(f"Нашли первое чётное: {num}")
        break
````

````python
# continue: пропускаем нечётные, печатаем чётные
for num in range(1, 11):
    if num % 2 != 0:
        continue    # нечётные пропускаем
    print(num)
````

````python
# else в цикле: срабатывает, если break не выполнился
numbers = [2, 4, 6, 8]

for num in numbers:
    if num % 2 != 0:
        print("Есть нечётное!")
        break
else:
    print("Нечётных нет — все числа чётные")
````

````python
# А теперь добавим нечётное число — else НЕ сработает
numbers = [2, 4, 7, 8]

for num in numbers:
    if num % 2 != 0:
        print(f"Есть нечётное: {num}!")
        break
else:
    print("Нечётных нет")
````

## 8. Практические примеры

````python
# Задача: найти сумму и количество чётных чисел в списке
numbers = [3, 8, 12, 5, 20, 7, 6]
count = 0
total = 0

for num in numbers:
    if num % 2 == 0:
        count += 1
        total += num

print(f"Чётных чисел: {count}")
print(f"Их сумма: {total}")
````

````python
# Задача: fizz-buzz (классика собеседований)
# Кратно 3 -> "Fizz", кратно 5 -> "Buzz", кратно обоим -> "FizzBuzz"
for i in range(1, 16):
    if i % 15 == 0:
        print("FizzBuzz")
    elif i % 3 == 0:
        print("Fizz")
    elif i % 5 == 0:
        print("Buzz")
    else:
        print(i)
````

````python
# Вложенные циклы: рисуем треугольник из звёздочек
height = 5
for row in range(1, height + 1):
    print("*" * row)
````

````python
# while + if: вычисляем факториал
n = 6
result = 1
i = 1

while i <= n:
    result *= i
    i += 1

print(f"{n}! = {result}")
````
