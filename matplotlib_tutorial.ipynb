# Matplotlib — туториал для Jupyter Notebook

**Matplotlib** — самая популярная библиотека Python для визуализации данных.
Этот ноутбук проведёт вас от первого графика до уверенного использования:
линейные графики, диаграммы рассеяния, столбчатые диаграммы, гистограммы,
кастомизация, сабплоты и сохранение в файл.

> Основной интерфейс — модуль `matplotlib.pyplot` (сокращённо `plt`).

В Jupyter обычно достаточно директивы `%matplotlib inline` — она встраивает
графики прямо в ноутбук (в современных версиях Jupyter это работает и без неё).

````python
import matplotlib.pyplot as plt
import numpy as np

%matplotlib inline
````

## 2. Первый график: `plt.plot`

`plot(x, y)` рисует линию по точкам. Если `x` не указан — используются индексы.

````python
x = np.linspace(0, 2 * np.pi, 100)
y = np.sin(x)

plt.plot(x, y)
plt.title("Синус")
plt.xlabel("x")
plt.ylabel("sin(x)")
plt.show()
````

## 3. Несколько линий и легенда

Каждый вызов `plt.plot` добавляет новую линию. Метки для легенды задаются
через `label`, легенда выводится через `plt.legend()`.

````python
x = np.linspace(0, 2 * np.pi, 100)

plt.plot(x, np.sin(x), label="sin(x)", linewidth=2)
plt.plot(x, np.cos(x), label="cos(x)", linestyle="--")
plt.plot(x, np.sin(2 * x), label="sin(2x)", linestyle=":", color="red")

plt.title("Три функции на одном графике")
plt.xlabel("x")
plt.ylabel("y")
plt.legend()
plt.grid(alpha=0.3)
plt.show()
````

### Стиль линий (формат-строка)

Первым аргументом после координат можно передать строку вида `"ro--"`:

| Символ | Цвет | Символ | Маркер | Символ | Линия |
|--------|------|--------|--------|--------|-------|
| `b` | синий | `.` | точка | `-` | сплошная |
| `g` | зелёный | `o` | круг | `--` | пунктир |
| `r` | красный | `s` | квадрат | `:` | точечная |
| `k` | чёрный | `^` | треугольник | `-.` | штрихпунктир |

Полный список: https://matplotlib.org/stable/api/_as_gen/matplotlib.pyplot.plot.html

## 4. Размер фигуры и ограничение осей

````python
x = np.linspace(0, 10, 200)

plt.figure(figsize=(10, 4))  # ширина, высота в дюймах
plt.plot(x, np.sin(x))
plt.xlim(0, 10)      # диапазон по оси X
plt.ylim(-1.5, 1.5)  # диапазон по оси Y
plt.show()
````

## 5. Диаграмма рассеяния: `plt.scatter`

Используется для отображения взаимосвязи между двумя переменными.
Размер и цвет точек можно сделать зависимыми от данных.

````python
rng = np.random.default_rng(42)
x = rng.normal(0, 1, 200)
y = 2 * x + rng.normal(0, 0.5, 200)
sizes = np.abs(y) * 100  # размер точки зависит от значения

plt.figure(figsize=(7, 5))
plt.scatter(x, y, s=sizes, c=y, cmap="viridis", alpha=0.7)
plt.colorbar(label="y")
plt.title("Диаграмма рассеяния")
plt.xlabel("x")
plt.ylabel("y")
plt.show()
````

## 6. Столбчатые диаграммы: `bar` и `barh`

````python
categories = ["Янв", "Фев", "Мар", "Апр", "Май"]
values = [12, 19, 8, 15, 23]

plt.bar(categories, values, color="steelblue")
plt.title("Столбчатая диаграмма")
plt.ylabel("Значение")
plt.show()
````

````python
# Сгруппированные столбцы
labels = ["A", "B", "C"]
series1 = [10, 15, 7]
series2 = [8, 12, 11]

x = np.arange(len(labels))
width = 0.35

plt.bar(x - width/2, series1, width, label="2025")
plt.bar(x + width/2, series2, width, label="2026")
plt.xticks(x, labels)
plt.legend()
plt.title("Сравнение двух рядов")
plt.show()
````

## 7. Гистограмма и boxplot

````python
data = rng.normal(100, 15, 1000)  # нормальное распределение

plt.hist(data, bins=30, edgecolor="black", alpha=0.7)
plt.title("Гистограмма")
plt.xlabel("Значение")
plt.ylabel("Частота")
plt.show()
````

````python
# Ящик с усами — медиана, квартили и выбросы
groups = [rng.normal(0, 1, 100), rng.normal(1, 0.5, 100), rng.normal(-0.5, 1.5, 100)]

plt.boxplot(groups, labels=["A", "B", "C"])
plt.title("Boxplot")
plt.ylabel("Значение")
plt.show()
````

## 8. Круговая и кольцевая диаграмма

````python
sizes = [35, 25, 20, 12, 8]
labels = ["Python", "SQL", "Java", "JS", "Прочее"]

plt.pie(sizes, labels=labels, autopct="%1.1f%%", startangle=90)
plt.title("Доли языков")
plt.show()
````

## 9. Сабплоты: несколько графиков рядом

`plt.subplots(nrows, ncols)` возвращает фигуру и массив осей (axes).
Дальше каждый subplot рисуется на своей оси — это рекомендуемый «объектный» стиль.

````python
fig, axes = plt.subplots(2, 2, figsize=(11, 8))
x = np.linspace(0, 2 * np.pi, 100)

axes[0, 0].plot(x, np.sin(x))
axes[0, 0].set_title("sin(x)")

axes[0, 1].plot(x, np.cos(x), color="orange")
axes[0, 1].set_title("cos(x)")

axes[1, 0].scatter(rng.random(50), rng.random(50))
axes[1, 0].set_title("scatter")

axes[1, 1].hist(rng.normal(0, 1, 500), bins=20, edgecolor="black")
axes[1, 1].set_title("hist")

fig.suptitle("Сетка 2×2", fontsize=16)
plt.tight_layout()
plt.show()
````

### Подграфики друг под другом — `plt.subplot`

````python
plt.figure(figsize=(8, 6))

plt.subplot(2, 1, 1)  # 2 строки, 1 колонка, 1-я ячейка
plt.plot(x, np.sin(x))
plt.title("Верхний график")

plt.subplot(2, 1, 2)
plt.plot(x, np.cos(x))
plt.title("Нижний график")

plt.tight_layout()
plt.show()
````

## 10. Настройка внешнего вида

### Цвета
- по имени: `"red"`, `"steelblue"`, `"tab:green"`
- по HEX: `"#ff5733"`
- по RGBA-кортежу: `(1.0, 0.2, 0.1, 0.5)`

### Глобальные настройки через `rcParams`

````python
plt.rcParams.update({
    "figure.figsize": (8, 5),
    "font.size": 12,
    "axes.grid": True,
    "grid.alpha": 0.3,
})

plt.plot(x, np.sin(x))
plt.title("График с глобальными настройками")
plt.show()
````

````python
# Готовые стили
print(plt.style.available)
plt.style.use("seaborn-v0_8-darkgrid")

plt.plot(x, np.sin(x))
plt.title('Стиль "seaborn-v0_8-darkgrid"')
plt.show()
````

## 11. Текст, аннотации и стрелки

````python
x = np.linspace(0, 2 * np.pi, 100)
y = np.sin(x)

plt.plot(x, y)
plt.annotate(
    "максимум",
    xy=(np.pi / 2, 1),
    xytext=(2.5, 0.3),
    arrowprops=dict(arrowstyle="->", color="red"),
    fontsize=12,
)
plt.text(4.5, -0.8, "plt.text() вставляет текст", fontsize=10, color="gray")
plt.title("Аннотации")
plt.show()
````

## 12. Оси с двумя шкалами (twinx)

````python
x = np.arange(1, 13)
temperature = [-5, -3, 2, 10, 17, 21, 23, 21, 15, 8, 1, -3]
precipitation = [45, 38, 40, 42, 50, 65, 80, 75, 60, 55, 50, 48]

fig, ax1 = plt.subplots()
ax1.bar(x, precipitation, color="lightblue", label="Осадки, мм")
ax1.set_xlabel("Месяц")
ax1.set_ylabel("Осадки, мм", color="tab:blue")

ax2 = ax1.twinx()  # вторая ось Y
ax2.plot(x, temperature, color="tab:red", marker="o", label="Температура, °C")
ax2.set_ylabel("Температура, °C", color="tab:red")

plt.title("Две шкалы на одном графике")
plt.show()
````

## 13. Логарифмические шкалы, ошибки и заливка

````python
x = np.linspace(1, 100, 50)
y = x ** 1.5

plt.loglog(x, y)  # логарифмические шкалы по обеим осям
plt.title("log-log график")
plt.show()
````

````python
x = np.arange(10)
y = np.random.default_rng(0).normal(5, 1, 10)
error = 0.5 + 0.3 * np.abs(y) / 5

plt.errorbar(x, y, yerr=error, fmt="o-", capsize=4)
plt.title("График с погрешностями")
plt.show()
````

````python
x = np.linspace(0, 10, 100)
y = np.sin(x)

plt.plot(x, y)
plt.fill_between(x, y - 0.2, y + 0.2, alpha=0.3, label="±0.2")
plt.legend()
plt.title("Заливка области между кривыми")
plt.show()
````

## 14. Дата-время на оси X

Для временных рядов используйте `matplotlib.dates` — даты разберутся и
подпишутся автоматически.

````python
import matplotlib.dates as mdates
from datetime import datetime, timedelta

dates = [datetime(2026, 1, 1) + timedelta(days=i) for i in range(90)]
values = np.cumsum(np.random.default_rng(1).normal(0, 1, 90))

plt.figure(figsize=(11, 4))
plt.plot(dates, values)
plt.gca().xaxis.set_major_formatter(mdates.DateFormatter("%d.%m"))
plt.gca().xaxis.set_major_locator(mdates.WeekdayLocator(interval=2))
plt.title("Временной ряд")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
````

## 15. Сохранение графика в файл

`savefig` поддерживает PNG, PDF, SVG, JPG и другие форматы.
`dpi` задаёт разрешение — важно для публикаций.

````python
x = np.linspace(0, 2 * np.pi, 100)

plt.figure(figsize=(6, 4))
plt.plot(x, np.sin(x), label="sin(x)")
plt.legend()
plt.title("График для сохранения")
plt.savefig("sine_plot.png", dpi=300, bbox_inches="tight")
plt.savefig("sine_plot.pdf", bbox_inches="tight")
plt.show()
````

## 16. Полезные приёмы и подводные камни

1. **Смешивание pyplot и объектного стиля.** Если используете `fig, ax = plt.subplots()`, обращайтесь к методам оси: `ax.set_title()`, `ax.set_xlabel()` — а не `plt.title()`.
2. **График не отображается** — убедитесь, что ячейка заканчивается на `plt.show()` или что включён inline-бэкенд.
3. **Перекрывающиеся подписи** — добавьте `plt.tight_layout()` перед `plt.show()`.
4. **Шрифты с кириллицей.** Если русский текст выглядит как квадратики:
   ```python
   plt.rcParams["font.family"] = "DejaVu Sans"  # поддерживает кириллицу
   ```
5. **Воспроизводимость** — фиксируйте seed у генератора случайных чисел (`np.random.default_rng(42)`).
